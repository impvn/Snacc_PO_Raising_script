```python
def login(User_id, Password):
  try:
    email_mobile_field = wait2.until(EC.presence_of_element_located((By.NAME, "username")))
    email_mobile_field.send_keys(User_id)
    password_field = wait2.until(EC.presence_of_element_located((By.ID, "id_password")))
    password_field.send_keys(Password)
    login_button = wait2.until(EC.element_to_be_clickable((By.XPATH, "//button[@type='submit']")))
    login_button.click()
    print("Login Successful")
  except Exception as e:
    print("Login inputs Not Available")

def select_suggestion(input_element, value):
   input_element.send_keys(value)
   try:
       suggestions = wait.until(EC.presence_of_all_elements_located((By.CSS_SELECTOR, "ul.md-autocomplete-suggestions li")))
       for suggestion in suggestions:
           suggestion_text = suggestion.text.strip()
           if value.lower() in suggestion_text.lower():
               suggestion.click()
               break
   except Exception as e:
       print(f"Suggestion not found for: {value} - {e}")

def retry(save_and_send_button,n):
   if n==0:
       return
   try:
       if save_and_send_button.is_enabled():
           driver.execute_script("arguments[0].click();", save_and_send_button)
           return
   except Exception as e:
       retry(save_and_send_button,n-1)
       print(f"Save and Send click failed. Execption : {e}")

def submit():
   try:
       save_and_send_button = driver.find_element(By.XPATH, "/html/body/md-content/section/ui-view/section/div/div/div/md-toolbar[2]/div/div[2]/button[2]/span")
       driver.execute_script("arguments[0].click();", save_and_send_button)
   except:
       retry(save_and_send_button,5)

def fill_form(vendor, pod, date, slot, items,count):
   try:
       driver.get("https://www.supplynote.in/orders/create")

       # Select Vendor
       vendor_input = wait.until(EC.presence_of_element_located((By.ID, "input-18")))
       select_suggestion(vendor_input, vendor)
       time.sleep(0.5)

       # Select POD (Delivery Address)
       pod_input = wait.until(EC.presence_of_element_located((By.ID, "input-21")))
       select_suggestion(pod_input, pod)

       # Fill Billing Address if Empty
       try:
           billing_address_input = driver.find_element(By.CSS_SELECTOR, 'input[placeholder="e.g Bagmane"]')
           billing_address_input.clear()
           billing_address_input.send_keys(pod)
           # time.sleep(1)
           suggestions = wait.until(EC.presence_of_all_elements_located((By.CSS_SELECTOR, "ul.md-autocomplete-suggestions li")))
           for suggestion in suggestions:
               if pod in suggestion.text:
                   suggestion.click()
                   break
       except NoSuchElementException:
           print("Billing address input not found. Skipping...")

       # Select Date
       date_inputs = driver.find_elements(By.CSS_SELECTOR, '.md-datepicker-input')
       if len(date_inputs) > 1:
           formatted_date = datetime.strptime(date, '%d/%m/%Y').strftime('%m/%d/%Y')
           date_inputs[1].send_keys(formatted_date)

       for item, quantity in items:
           product_input = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, 'input[placeholder="Search Product to add"]')))
           product_input.send_keys(item)
           time.sleep(1) # try with 3 second of sleep in case of repeated failures.

           suggestions = wait.until(EC.presence_of_all_elements_located((By.CSS_SELECTOR, "ul.md-autocomplete-suggestions li")))
           for suggestion in suggestions:
               suggestion_text = suggestion.text.strip()
               if item.lower() in suggestion_text.lower():
                   suggestion.click()
                   break

           quantity_input = driver.find_element(By.CSS_SELECTOR, 'input[ng-model="vm.itemToAdd.quantity"]')
           quantity_input.send_keys(quantity)
           quantity_input.send_keys(Keys.TAB)

           add_button = wait.until(EC.element_to_be_clickable((By.XPATH, '/html/body/md-content/section/ui-view/section/md-content/md-card/md-card-content/div[2]/table[1]/tbody/tr/td[10]/div/button[1]/span')))
           add_button.click()
           count+=1

           if count == len(items):
               wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR, 'input[placeholder="Search Product to add"]')))
               return count

   except Exception as e:
       print(f"Error filling form for {vendor}, {pod}, {slot}")
       return count

def start_raising(MAX_RETRIES,Successful,failed, grouped_data):
  for (vendor, pod, date, template), group in grouped_data:
   items = list(zip(group['SKUcode'], group['Quantity']))
   count = 0
   attempt = 0

   print(f"Processing group: Date={date}, Vendor={vendor}, POD={pod}, Slot={template}")

   while attempt < MAX_RETRIES and count < len(items):
       count = fill_form(vendor, pod, date, template, items, 0)
       print(f"Attempt {attempt + 1}: Items added: {count}/{len(items)}")

       if count == len(items):
           break
       attempt += 1
       time.sleep(2)

   if count == len(items):
       submit()
       print("Form submitted successfully")
       Successful.append([date,vendor,pod,template])
   else:
       print(f"Failed to add all items after {MAX_RETRIES} attempts: {count}/{len(items)}")
       failed.append([date,vendor,pod,template])

def failed_groups_status(failed):
  worksheet = client.open("BLR-Forecasting-Sheet-POs-Pawan-3-New-Template").worksheet("PO_Status")
  Failed_df = pd.DataFrame(failed, columns=['Date','Vendor','Location','Slot'])
  failed_data = pd.merge(Failed_df, df, how='left', on=['Date','Vendor','Location','Slot'])
  worksheet.clear()
  set_with_dataframe(worksheet, failed_data)

def process_failed_groups(failed):
  Failed_df = pd.DataFrame(failed, columns=['Date','Vendor','Location','Slot'])
  failed_data = pd.merge(Failed_df, df, how='left', on=['Date','Vendor','Location','Slot'])
  failed_data.columns = failed_data.columns.str.strip()
  failed_grouped_data = failed_data.groupby(['Vendor', 'Location', 'Date', 'Slot'])
  MAX_RETRIES = 2
  Successful = []
  failed = []
  start_raising(MAX_RETRIES, Successful,failed, failed_grouped_data)
  failed_groups_status(failed)
  return failed
```
